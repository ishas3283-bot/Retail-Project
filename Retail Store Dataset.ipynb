# Retail Store Sales

### What are we trying to Find?
Through this dataset of retail store, we would perform cleaning of data which is priortise over any process before jumping to find queries for business insights. Therefore, the workflow of this project starts from Python Library - Pandas, for easy and smart cleaning of data.

### Setting up the Library

```python
import numpy as np
import pandas as pd

store = pd.read_csv('retail_store_sales.csv')
```

```python
store.dropna(inplace=True)   #Delete the missing values
```

```python
store['Transaction Date'] = pd.to_datetime(store['Transaction Date'])   #Convert String to Date format in Transaction ID Column
```

```python
store.drop_duplicates()
```

```python
store.fillna({'Price Per Unit':45}, inplace=True)  #Fill missing values of column(Price per unit) = 45, if any missing values exist
```

```python
store.replace('Digital Wallet', 'Online/UPI', inplace=True) 
store.replace('Credit Card', 'Debit/Credit', inplace=True) #Replacing the name 
```

```python
store.to_csv('cleaned_dataset', index = True)  #Saving the file format in local device
```
